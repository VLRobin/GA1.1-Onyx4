```python
def add(a, b):
    return a + b
```

```python
result1 = add(3, 7)
print("The result is:", result1)

result2 = add(9, 2)
print("The result is:", result2)

result3 = add(29, 53)
print("The result is:", result3)
```

```python
def sum_list(values):
    list = []
    for i in range(len(values)):
        list.append(values[i])
    return sum(list)
print(sum_list([result1, result2, result3]))
```
